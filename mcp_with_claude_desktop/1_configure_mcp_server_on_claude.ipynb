#### Add Time mcp server on claude

Go to settings -> Developer -> Edit config. Restart

If you have enabled developer mode (help -> enable developer mode), then you can go to 

Menu-> developer -> reload mcp configuration

Once uv is installed, on windows cmd prompt get path by 

```sh
where uv
```

Use that path and add below json in config

```json
{
  "mcpServers": {
    "time": {
      "command": "C:\\Users\\PriyabrataPanigrahi\\.local\\bin\\uvx.exe",
      "args": ["mcp-server-time"]
    }
  }
}
```

#### File system MCP server

Find out where is node in windows command prompt

```sh
where npx
```

Give the path of npx


Add the file system mcp server

```json
{
  "mcpServers": {
    "time": {
      "command": "C:\\Users\\PriyabrataPanigrahi\\.local\\bin\\uvx.exe",
      "args": ["mcp-server-time"]
    },
    "filesystem": {
      "command": "C:\\Program Files\\nodejs\\npx.cmd",
      "args": [
        "-y",
        "@modelcontextprotocol/server-filesystem",
        "C:\\Users\\PriyabrataPanigrahi\\Downloads\\work\\sas"
      ]
    }
  }
}
```

Restart Claude to see the effect
